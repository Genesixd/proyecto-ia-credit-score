## PREPROCESAMIENTO BASICO

```python
import pandas as pd

df = pd.read_csv("credit_score.csv")
df.head()
```

```python
#Revisar valores nulos por columna
#Esto te muestra qué columnas tienen más NaN
df.isnull().sum().sort_values(ascending=False)
```

> **Nota:** Se excluyeron `ID`, `Customer_ID`, `Name`, `SSN` y `Month` por ser identificadores/datos sensibles; `Month` se deja fuera por ahora para simplificar el modelo.



```python
cols_drop = ["ID", "Customer_ID", "Name", "SSN", "Month"]
df = df.drop(columns=cols_drop)
df.head()
```

```python
#Detectar tipos de variables (numéricas vs categóricas)
df.dtypes
```

```python

#ver listas
num_cols = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
cat_cols = df.select_dtypes(include=["object"]).columns.tolist()

num_cols, cat_cols
```

> **Nota:** En la limpieza básica se reemplazan valores raros como `"_"`, `"10000"`, `"NA"`, `"NM"`, etc. por `NaN`, tratándolos como datos faltantes.


```python
import numpy as np

# Lista de tokens que consideraremos como "faltantes"
rare_tokens = ["_", "__", "NA", "NM", "Not Specified", "__10000__", "!", "!@9#%8", "-","_______"]

df = df.replace(rare_tokens, np.nan)
```

```python
#Imputar numéricas con la mediana
num_cols = df.select_dtypes(include=["int64", "float64"]).columns

for col in num_cols:
    med = df[col].median()
    df[col] = df[col].fillna(med)
```

```python
#Imputar categóricas con la moda
cat_cols = df.select_dtypes(include=["object"]).columns

for col in cat_cols:
    mode = df[col].mode(dropna=True)
    if len(mode) > 0:
        df[col] = df[col].fillna(mode[0])
    else:
        df[col] = df[col].fillna("Unknown")
```

```python
#Verificar que ya no queden NaN
df.isnull().sum().sort_values(ascending=False).head(15)
```

```python
#Guardar una versión limpia para usar luego
df.to_csv("credit_score_clean.csv", index=False)
```
